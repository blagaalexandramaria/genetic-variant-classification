# Final Model

## Objective

This notebook trains the final pathogenicity prediction model using the most informative genomic and clinical features identified during baseline experimentation.

### Features

- Chromosome
- GeneSymbol
- mutation_type
- OriginSimple
- ReviewStatus
- ref_length
- alt_length
- length_difference

### Model

Random Forest Classifier with:

- OneHotEncoder
- Class balancing
- Increased number of trees

The goal is to obtain the best predictive performance while maintaining interpretability.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    classification_report,
    confusion_matrix
)
```

```python
df = pd.read_csv(

    "../data/processed/clinvar_features.csv"

)

df.shape
```

```python
features = [
    "Chromosome",
    "GeneSymbol",
    "mutation_type",
    "OriginSimple",
    "ReviewStatus",
    "ref_length",
    "alt_length",
    "length_difference"
]

target = "target"
```

```python
df_final = df[
    features + [target]
].dropna().copy()

df_final.shape
```

```python
categorical_features = [
    "Chromosome",
    "GeneSymbol",
    "mutation_type",
    "OriginSimple",
    "ReviewStatus"
]

numeric_features = [
    "ref_length",
    "alt_length",
    "length_difference"
]

for col in categorical_features:
    df_final[col] = df_final[col].astype(str)
```

```python
sample_final = df_final.sample(
    n=50000,
    random_state=42
).copy()

sample_final.shape
```

```python
X = sample_final[features]

y = sample_final[target]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_features
        ),
        (
            "num",
            "passthrough",
            numeric_features
        )
    ]
)
```

```python
final_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),

        ("classifier",
         RandomForestClassifier(
             n_estimators=300,
             max_depth=20,
             class_weight="balanced",
             random_state=42,
             n_jobs=-1
         ))
    ]
)
```

```python
final_model.fit(
    X_train,
    y_train
)
```

```python
predictions = final_model.predict(
    X_test
)
```

```python
print(
    classification_report(
        y_test,
        predictions
    )
)
```

```python
cm = confusion_matrix(
    y_test,
    predictions
)

print(cm)
```

```python
rf = final_model.named_steps["classifier"]

importances = rf.feature_importances_
```

```python
feature_names = (
    final_model.named_steps["preprocessor"]
    .get_feature_names_out()
)
```

```python
importance_df = pd.DataFrame({
    "feature": feature_names,
    "importance": importances
})

importance_df = importance_df.sort_values(
    by="importance",
    ascending=False
)

importance_df.head(20)
```

```python
import matplotlib.pyplot as plt

top20 = importance_df.head(20)

plt.figure(figsize=(10,6))

plt.barh(
    top20["feature"],
    top20["importance"]
)

plt.xlabel("Importance")
plt.ylabel("Feature")
plt.title("Top 20 Most Important Features")

plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```
