```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/processed/clinvar_grch38_clean.csv")

df.shape
```

```python
df[[
    "Type",
    "ClinicalSignificance",
    "target",
    "Chromosome",
    "ReferenceAllele",
    "AlternateAllele",
    "ReferenceAlleleVCF",
    "AlternateAlleleVCF"
]].head()
```

```python
df["ref"] = df["ReferenceAlleleVCF"].astype(str)
df["alt"] = df["AlternateAlleleVCF"].astype(str)
```

```python
df["ref_length"] = df["ref"].str.len()
df["alt_length"] = df["alt"].str.len()
df["length_difference"] = df["alt_length"] - df["ref_length"]
```

```python
df["is_snv"] = ((df["ref_length"] == 1) & (df["alt_length"] == 1)).astype(int)

df["is_deletion"] = (df["alt_length"] < df["ref_length"]).astype(int)

df["is_insertion"] = (df["alt_length"] > df["ref_length"]).astype(int)
```

```python
df[[
    "ref",
    "alt",
    "ref_length",
    "length_difference",
    "is_snv",
    "is_deletion",
    "is_insertion",
    "target"
]].head(20)
```

```python
df[["ref_length", "alt_length", "length_difference"]].describe()
```

```python
df[["is_snv", "is_deletion", "is_insertion"]].sum()
```

```python
df.to_csv(
    "../data/processed/clinvar_features.csv",
    index=False
)
```

```python
df = df[
    (df["ref"] != "na") &
    (df["alt"] != "na")
].copy()
```

```python
df.shape
```

```python
df["mutation_type"] = "other"

df.loc[
    (df["ref_length"] == 1) &
    (df["alt_length"] == 1),
    "mutation_type"
] = "SNV"

df.loc[
    df["alt_length"] > df["ref_length"],
    "mutation_type"
] = "Insertion"

df.loc[
    df["alt_length"] < df["ref_length"],
    "mutation_type"
] = "Deletion"
```

```python
df["mutation_type"].value_counts()
```

```python
df.to_csv(
    "../data/processed/clinvar_features.csv",
    index=False
)
```
