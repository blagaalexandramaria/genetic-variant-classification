```python
import pandas as pd

df = pd.read_csv(
    "../data/raw/variant_summary.txt",
    sep="\t",
    low_memory=False
)
```

```python
selected_classes = [
    "Benign",
    "Likely benign",
    "Pathogenic",
    "Likely pathogenic"
]

df_clean = df[
    df["ClinicalSignificance"].isin(selected_classes)
].copy()
```

```python
df_clean.shape
```

```python
df_clean["ClinicalSignificance"].value_counts()
```

```python
pathogenic_classes = [
    "Pathogenic",
    "Likely pathogenic"
]

df_clean["target"] = (
    df_clean["ClinicalSignificance"]
    .isin(pathogenic_classes)
    .astype(int)
)
```

```python
df_clean["target"].value_counts()
```

```python
len(df_clean)
```

```python
df_clean["VariationID"].nunique()
```

```python
df_clean["GeneSymbol"].nunique()
```

```python
pathogenic_df = df_clean[
    df_clean["target"] == 1
]

pathogenic_df["GeneSymbol"].value_counts().head(20)
```

```python
benign_df = df_clean[
    df_clean["target"] ==0
]

benign_df["GeneSymbol"].value_counts().head(20)
```

```python
import matplotlib.pyplot as plt

df_clean["ClinicalSignificance"]\
.value_counts()\
.plot(kind="bar")

plt.title("Clinical Significance Distribution")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```
